####Почему Ф. М. Достоевский?

Ф. М. Достоевский – ключевая фигура для любого исследования русской психологической прозы. Ф. М. Достоевский заложил основу так называемой «диалектики души», которая впоследствии стала краеугольным камнем литературы рубежа XIX–XX веков. Именно он впервые в русской словесности сделал внутренний мир героя, его болезненные состояния, нравственные метания и экзистенциальные кризисы самостоятельным сюжетом, а не только средством характеристики персонажа.

Ф. М. Достоевский, очевидно, оказал влияние на Л. Н. Андреева: в конце жизни последний писал, что считает себя последователем и учеником Ф. М. Достоевского (Пращерук Н. В. Несостоявшийся диалог: Ф. Достоевский и Л. Андреев //Эволюция форм художественного сознания в русской литературе. Вып. 4 Екатеринбург, 2014.).

Однако стилистически авторы принципиально расходятся: если Ф. М. Достоевский тяготеет к длинным, синтаксически сложным периодам, к полилогу и философским диалогам, то Л. Н. Андреев использует экспрессивный, рваный синтаксис, гротеск и гиперболу. Этот контраст свидетельствует об эволюции поэтики безумия от классического психологизма к экспрессионизму начала XX века и подчеркивает новаторство Андреева в модернистской литературе.

```python
import requests
from bs4 import BeautifulSoup
import time
import os
import re
from urllib.parse import urljoin

# Расширенный список стоп-фраз для удаления служебных строк
SERVICE_PHRASES = [
    'Lib.ru/Классика',
    'Регистрация',
    'Найти',
    'Обсуждения',
    'Новинки',
    'Обзоры',
    'Помощь',
    'Достоевский Федор Михайлович',
    "Достоевский",
    "yes@lib.ru",
    "Аннотация",
    "Федор",
    "Михайлович",
    "Федор Михайлович",
    "Год",
    "Рассказ",
    "римечания",
    "П",
    'Обновлено:',
    'Статистика.', ")", "(", "[", ']',
    'Повесть',
    'Проза',
    'Повести и рассказы',
    'FB2',
    '6.06*72',
    'оценка', 'комментари', 'рейтинг', 'обсудить',
    'читать', 'скачать', 'просмотров', 'голосов',
    'дата', 'источник', 'ссылка', 'назад', 'вверх',
    'содержание', 'оглавление', 'предисловие', 'примечание', 'сноска', 'С.', 'Оригинал здесь:',
    # Оценки (отзывы)
    'шедевр', 'замечательно', 'очень хорошо', 'хорошо',
    'нормально', 'Не читал', 'терпимо', 'посредственно',
    'плохо', 'очень плохо'
]

def is_service_line(line):
    """Проверяет, является ли строка служебной (реклама, оценки, навигация и т.д.)"""
    line_lower = line.lower()
    # Если строка содержит хотя бы одну стоп-фразу, считаем её служебной
    for phrase in SERVICE_PHRASES:
        if phrase.lower() in line_lower:
            return True
    # Удаляем строки, состоящие только из цифр, знаков и пробелов
    if re.match(r'^[\d\s\.,;:!?\-]+$', line):
        return True
    # Удаляем строки, которые начинаются с "www." или "http"
    if re.match(r'^(www\.|http|ftp)', line_lower):
        return True
    # Удаляем строки, которые содержат только одно слово из списка оценок (например, "шедевр")
    if re.match(r'^(шедевр|замечательно|очень хорошо|хорошо|нормально|не читал|терпимо|посредственно|плохо|очень плохо)$', line_lower):
        return True
    return False

# 1. Парсинг ссылок только из раздела "Повести и рассказы"
def get_text_links_only_stories(base_url):
    response = requests.get(base_url)
    response.encoding = 'cp1251'
    soup = BeautifulSoup(response.text, 'html.parser')

    start_elem = soup.find('a', attrs={'name': 'gr2'})
    if start_elem is None:
        for elem in soup.find_all(['h3', 'b', 'strong', 'p']):
            if 'Повести и рассказы' in elem.get_text():
                start_elem = elem
                break

    if start_elem is None:
        print("Раздел не найден. Собираем все ссылки (может быть лишнее).")
        return get_all_text_links(base_url)

    all_elements = soup.find_all()
    try:
        start_index = all_elements.index(start_elem)
    except ValueError:
        return get_all_text_links(base_url)

    links = []
    for elem in all_elements[start_index + 1:]:
        if elem.name == 'a' and elem.get('name') and elem.get('name').startswith('gr') and elem.get('name') != 'gr2':
            break
        if elem.name in ['h3', 'b', 'strong'] and any(
            word in elem.get_text() for word in ['Романы', 'Публицистика', 'Критика', 'Дневник']
        ):
            break
        if elem.name == 'a' and elem.get('href'):
            href = elem['href']
            if href.startswith('text_') and href.endswith('.shtml'):
                links.append(urljoin(base_url, href))

    links = list(set(links))
    print(f"Найдено {len(links)} ссылок на тексты в разделе 'Повести и рассказы'.")
    return links

def get_all_text_links(base_url):
    response = requests.get(base_url)
    response.encoding = 'cp1251'
    soup = BeautifulSoup(response.text, 'html.parser')
    links = []
    for a in soup.find_all('a', href=True):
        href = a['href']
        if href.startswith('text_') and href.endswith('.shtml'):
            links.append(urljoin(base_url, href))
    return list(set(links))

# 2. Парсинг текста со страницы с очисткой от верстки
def parse_text_page(url):
    try:
        response = requests.get(url)
        response.encoding = 'cp1251'
        soup = BeautifulSoup(response.text, 'html.parser')

        # Ищем основной текст в теге <pre>
        text_block = soup.find('pre')
        if text_block:
            raw_text = text_block.get_text(separator='\n')
        else:
            # Если <pre> нет, удаляем мусорные теги и берём остальное
            for tag in soup(['script', 'style', 'nav', 'header', 'footer', 'aside']):
                tag.decompose()
            raw_text = soup.get_text(separator='\n')

        # Разбиваем на строки и фильтруем
        lines = raw_text.splitlines()
        clean_lines = []
        for line in lines:
            stripped = line.strip()
            if not stripped:
                continue
            if is_service_line(stripped):
                continue
            clean_lines.append(stripped)

        # Собираем обратно в единый текст
        clean_text = '\n'.join(clean_lines)
        return clean_text

    except Exception as e:
        print(f"Ошибка при загрузке {url}: {e}")
        return None

# 3. Сохранение отдельного файла
def save_text(url, text, output_dir='dostoevsky_stories'):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    filename_raw = url.split('/')[-1].replace('.shtml', '')
    safe_filename = re.sub(r'[<>:"/\\|?*]', '_', filename_raw)
    if len(safe_filename) > 200:
        safe_filename = safe_filename[:200]
    full_path = os.path.join(output_dir, safe_filename + '.txt')
    with open(full_path, 'w', encoding='utf-8') as f:
        f.write(text)
    print(f"Сохранено: {full_path}")
    return full_path

# 4. Объединение всех файлов в один
def merge_all_texts(input_dir='dostoevsky_stories', output_file='dostoevsky_all.txt'):
    if not os.path.exists(input_dir):
        print(f"Папка {input_dir} не найдена.")
        return None

    files = [f for f in os.listdir(input_dir) if f.endswith('.txt')]
    files.sort()

    if not files:
        print("В папке нет .txt файлов.")
        return None

    with open(output_file, 'w', encoding='utf-8') as out:
        for idx, filename in enumerate(files):
            filepath = os.path.join(input_dir, filename)
            with open(filepath, 'r', encoding='utf-8') as inf:
                content = inf.read()
            out.write(f"=== {filename} ===\n")
            out.write(content)
            if idx < len(files) - 1:
                out.write("\n\n---\n\n")

    print(f"Все файлы объединены в {output_file}")
    return output_file


# 5. Основной процесс
def main():
    base_url = 'http://az.lib.ru/d/dostoewskij_f_m/'
    text_links = get_text_links_only_stories(base_url)

    if not text_links:
        print("Ссылки не найдены. Проверьте структуру сайта.")
        return

    for url in text_links:
        print(f"Обработка: {url}")
        text = parse_text_page(url)
        if text:
            save_text(url, text)
        else:
            print(f"Не удалось получить текст с {url}")
        time.sleep(1)

    merged_file = merge_all_texts()


if __name__ == '__main__':
    main()
```

#### Почему В. М. Гаршин?
В. М. Гаршин – младший современник Ф. М. Достоевского и предшественник Л. Н. Андреева, чьё творческое наследие часто становится материалом в исследованиях о дискурсе / поэтике безумия. (Ссылки на работы: например.: Фролова О. В. Мотив безумия в рассказе Вс. М. Гаршина" Красный цветок" //Культура и текст. – 2019. – №. 1 (36). – С. 19-27. Шорина Э. В. Интертекстуальный дискурс темы безумия в рассказе ВМ Гаршина" Красный цветок" //Филологический аспект. – 2015. – №. 3. – С. 1-1.)

Большинство исследователей сосредотачивается на рассказе "Красный цветок". В этой же работе анализируется весь корпус малой прозы В. М. Гаршина. В корпус включаются повести, сказки и рассказы писателя – всего 19 текстов.

В. М. Гаршин, с одной стороны, сам испытал влияние психологической прозы Ф. М. Достоевского, с другой – его творчество уже содержит элементы экспрессионистской прозы рубежа XIX-XX веков, которые в полной мере разовьёт Л. Н. Андреев. Таким образом, он эстетически и хронологически выступает "мостом" между нашим основным объектом исследования, Л. Н. Андреевым, и самым влиятельным автором в теме описания психологии и безумия, Ф. М. Достоевским.

```python
import requests
from bs4 import BeautifulSoup
import time
import os
import re
from urllib.parse import urljoin
import pandas as pd

# НАСТРОЙКИ
BASE_URL = "http://az.lib.ru/g/garshin_w_m/"
SAVE_FOLDER = "garshin_malaya_proza"
os.makedirs(SAVE_FOLDER, exist_ok=True)

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}



def get_malaya_proza_links(base_url):
    print("Ищем разделы малой прозы...")
    response = requests.get(base_url, headers=headers)
    response.encoding = 'windows-1251'
    soup = BeautifulSoup(response.text, 'html.parser')

    links = []
    capture = False

    for elem in soup.find_all(['h3', 'b', 'strong', 'p', 'a']):
        text = elem.get_text(strip=True)

        if any(x in text for x in ['Рассказы и повесть', 'Сказки']):
            print(f"Нашли раздел: {text}")
            capture = True
            continue

        if any(x in text for x in ['Публицистика', 'Переписка', 'Критика', 'Поэзия', 'Драматургия', 'О творчестве']):
            if capture:
                print("Конец нужного раздела.")
            capture = False

        if capture and elem.name == 'a' and elem.get('href'):
            href = elem['href']
            if href.startswith('text_') and href.endswith('.shtml'):
                full_url = urljoin(base_url, href)
                title = elem.get_text(strip=True)
                if title and len(title) > 1:
                    links.append((title, full_url))

    # Убираем дубли
    unique_links = []
    seen = set()
    for title, url in links:
        if url not in seen:
            seen.add(url)
            unique_links.append((title, url))

    print(f"Найдено {len(unique_links)} произведений малой прозы.")
    return unique_links


def parse_text_page(url):
    try:
        response = requests.get(url, headers=headers)
        response.encoding = 'windows-1251'
        soup = BeautifulSoup(response.text, 'html.parser')

        text_block = soup.find('pre')
        if text_block:
            text = text_block.get_text(separator='\n')
        else:
            for tag in soup(['script', 'style', 'nav', 'header', 'footer', 'aside']):
                tag.decompose()
            text = soup.get_text(separator='\n')

        lines = [line.strip() for line in text.splitlines() if line.strip()]
        return '\n'.join(lines)
    except Exception as e:
        print(f"Ошибка загрузки: {e}")
        return None


# ЗАПУСК
text_links = get_malaya_proza_links(BASE_URL)

data = []
for i, (title, url) in enumerate(text_links, 1):
    print(f"[{i:2d}/{len(text_links)}] Скачиваем → {title}")

    text = parse_text_page(url)
    if text and len(text) > 500:   # защита от пустых страниц
        filename = url.split('/')[-1].replace('.shtml', '.txt')
        safe_filename = re.sub(r'[<>:"/\\|?*]', '_', filename)
        full_path = os.path.join(SAVE_FOLDER, safe_filename)

        with open(full_path, 'w', encoding='utf-8') as f:
            f.write(text)

        data.append({
            'title': title,
            'url': url,
            'length': len(text),
            'filename': safe_filename
        })

        time.sleep(1.1)
    else:
        print(f"   Ошибка: {e}")

# СОХРАНЕНИЕ РЕЗУЛЬТАТОВ
df = pd.DataFrame(data)
df.to_excel('Гаршин_малая_проза.xlsx', index=False)


print("\n=== ВСЁ ГОТОВО ===")
print(f"Скачано: {len(df)} произведений")
print(f"Папка с текстами: {SAVE_FOLDER}")
print("Также сохранен Excel файл")
```

Далее скрипт, очищающий получившиеся файлы от верстки сайта и объединяющий все тексты в один файл. В тексте останутся пометы от редакторов/издателей, но вся очевидно техническая информация будет удалена.

```python
import os
import re


FOLDER = "garshin_malaya_proza"
OUTPUT_FILE = "garshin_all_clean.txt"

def soft_clean(lines, filename):
    cleaned = []
    started = False

    for line in lines:
        stripped = line.strip()
        if not stripped and started:
            cleaned.append(line)
            continue
        if not stripped:
            continue

        # Начало текста
        if not started:
            if (len(stripped) > 55 and
                not any(x in stripped.lower() for x in
                    ['lib.ru', 'скачать', 'оценка', 'комментари', 'статистика',
                     'ваша оценка', 'оставить комментарий', 'обновлено', 'год:'])):

                started = True
                cleaned.append(line)
                continue
            else:
                continue

        # Убираем блок с "Оставить комментарий", "Ваша оценка" и т.д.
        if "Оставить комментарий" in stripped or "Ваша оценка:" in stripped:
            # Пропускаем весь этот блок (обычно 15-20 строк)
            continue

        if any(x in stripped for x in ["Год:", "Обновлено:", "Статистика.", "Оценка:"]):
            if len(cleaned) < 20:   # если текст ещё только начинается
                continue

        # Пропускаем OCR и правки
        if any(x in stripped.lower() for x in ['ocr', 'spellcheck', 'правка', 'доп.', 'печатается по']):
            if len(cleaned) < 15:
                continue

        # Конец файла
        if any(x in stripped for x in ["Книжная полка", "rusf.ru", "Связаться с программистом", "Комментарии:"]):
            break

        cleaned.append(line)

    final_text = ''.join(cleaned).strip()
    final_text = re.sub(r'\n{3,}', '\n\n', final_text)

    return final_text


# Запуск
print("Запускаем очистку...\n")

files = [f for f in os.listdir(FOLDER) if f.endswith('.txt')]
files.sort()

with open(OUTPUT_FILE, 'w', encoding='utf-8') as big_file:
    for i, filename in enumerate(files, 1):
        filepath = os.path.join(FOLDER, filename)

        with open(filepath, 'r', encoding='utf-8') as f:
            lines = f.readlines()

        cleaned_text = soft_clean(lines, filename)

        title = filename.replace('.txt', '').replace('_', ' ')

        big_file.write(f"{'='*85}\n")
        big_file.write(f"{i:2d}. {title}\n")
        big_file.write(f"{'='*85}\n\n")
        big_file.write(cleaned_text)
        big_file.write("\n\n\n\n")

        # Обновляем отдельный файл
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write(cleaned_text)

        print(f"[{i:2d}/{len(files)}] ✓ {title}")

print("\n" + "="*70)
print("ГОТОВО!")
print(f"Большой файл: {OUTPUT_FILE}")
print(f"Обработано: {len(files)} произведений")
print("="*70)
```

###Предобработка (токенизация, лемматизация, удаление стоп-слов)

```python
# Установка Natasha (если ещё не установлена)
!pip install natasha

import re
import os
import nltk
from nltk.corpus import stopwords
from natasha import Segmenter, MorphVocab, NewsEmbedding, NewsMorphTagger, Doc

# Загрузка стоп-слов
try:
    stopwords.words('russian')
except LookupError:
    nltk.download('stopwords')
russian_stopwords = set(stopwords.words('russian')) | {"это", "весь", "очень", "который", 'свой', "мочь", "сказать", "знать", "говорить", "что-то", "наш", "стать", "м", "петрович", "ваш", "идти", "иванович", "то", "что", "пойти", "-то", "фома", "франц", "какой", "какой-то", "николаевич"}

# Загрузка Natasha
segmenter = Segmenter()
morph_vocab = MorphVocab()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)

def preprocess_file(input_file, output_lemmas, output_clean):
    """
    Предобработка текстового файла: лемматизация и удаление стоп-слов.
    Результаты сохраняются в два файла: все леммы и предложения из лемм.
    """
    if not os.path.exists(input_file):
        print(f"Файл {input_file} не найден в /content/")
        return

    with open(input_file, 'r', encoding='utf-8') as f:
        raw_text = f.read()
    raw_text = re.sub(r'\s+', ' ', raw_text).strip()

    doc = Doc(raw_text)
    doc.segment(segmenter)
    doc.tag_morph(morph_tagger)
    for token in doc.tokens:
        token.lemmatize(morph_vocab)

    all_lemmas = []
    clean_sentences = []
    for sent in doc.sents:
        sent_lemmas = []
        for token in sent.tokens:
            if token.pos in ('PUNCT', 'SYM', 'X'):
                continue
            lemma = token.lemma
            if lemma.lower() in russian_stopwords:
                continue
            sent_lemmas.append(lemma)
        if sent_lemmas:
            all_lemmas.extend(sent_lemmas)
            clean_sentences.append(' '.join(sent_lemmas))

    with open(output_lemmas, 'w', encoding='utf-8') as f:
        f.write(' '.join(all_lemmas))
    print(f"Леммы сохранены в {output_lemmas} (всего {len(all_lemmas)})")

    with open(output_clean, 'w', encoding='utf-8') as f:
        f.write('\n'.join(clean_sentences))
    print(f"Очищенный текст сохранён в {output_clean} ({len(clean_sentences)} предложений)")

# Применяем к трём корпусам

# Достоевский
preprocess_file('dostoevsky_all.txt',
                'dostoevsky_lemmas.txt',
                'dostoevsky_clean.txt')

# Андреев
preprocess_file('/content/andreev_all.txt',
                'andreev_lemmas.txt',
                'andreev_clean.txt')

# Гаршин
preprocess_file('garshin_all_clean.txt',
                'garshin_lemmas.txt',
                'garshin_clean.txt')
```

###Расчет средней длины предложения

```python
import os
import re

def avg_sentence_length(input_file, author_name):
    """
    Вычисляет среднюю длину предложения (в словах) по тексту.
    Предложения выделяются по знакам . ! ?
    """
    if not os.path.exists(input_file):
        print(f"{input_file} не найден. Сначала загрузите файл.")
        return

    with open(input_file, 'r', encoding='utf-8') as f:
        text = f.read()

    # Разбиваем текст на предложения по . ! ? (с учётом последующего пробела или конца строки)
    # Используем lookbehind, чтобы сохранить разделители? Нам они не нужны, только текст предложений.
    sentences = re.split(r'[.!?]\s+', text)

    # Удаляем пустые или слишком короткие фрагменты
    sentences = [s.strip() for s in sentences if len(s.strip()) > 3]

    if not sentences:
        print("Нет предложений для анализа.")
        return

    # Считаем количество слов в каждом предложении (буквенные последовательности, включая русские и английские)
    word_counts = []
    for sent in sentences:
        words = re.findall(r'[а-яёa-z]+', sent, re.IGNORECASE)
        if words:
            word_counts.append(len(words))

    if not word_counts:
        print("Не удалось выделить слова в предложениях.")
        return

    avg = sum(word_counts) / len(word_counts)
    print(f"Средняя длина предложения в рассказах {author_name}: {avg:.2f} слов")

# Применяем ко всем трём авторам
print("--- Расчёт средней длины предложения ---")
avg_sentence_length('dostoevsky_all.txt', 'Достоевского')
avg_sentence_length('garshin_all_clean.txt', 'Гаршина')
avg_sentence_length('andreev_all.txt', 'Андреева')
```

```python
import os

def avg_word_length(lemmas_file, author_name):
    """
    Вычисляет среднюю длину слова (в символах) по файлу с леммами.
    """
    if not os.path.exists(lemmas_file):
        print(f"{lemmas_file} не найден. Сначала выполните предобработку.")
        return
    with open(lemmas_file, 'r', encoding='utf-8') as f:
        lemmas = f.read().split()
    if not lemmas:
        print(f"В файле {lemmas_file} нет лемм.")
        return
    total = sum(len(w) for w in lemmas)
    avg = total / len(lemmas)
    print(f"Средняя длина слова в рассказах {author_name}: {avg:.2f} символов")

# Применяем ко всем трём авторам
print("--- Расчёт средней длины слова ---")
avg_word_length('dostoevsky_lemmas.txt', 'Достоевского')
avg_word_length('garshin_lemmas.txt', 'Гаршина')
avg_word_length('andreev_lemmas.txt', 'Андреева')
```

###Расчет дисперсии предложений

```python
import os
import statistics

def variance_sentence_length(clean_file, author_name):
    """
    Вычисляет дисперсию длины предложений (в словах) по очищенному тексту.
    """
    if not os.path.exists(clean_file):
        print(f"{clean_file} не найден. Сначала выполните предобработку.")
        return
    with open(clean_file, 'r', encoding='utf-8') as f:
        lines = f.read().splitlines()
    lengths = [len(line.split()) for line in lines if line.strip()]
    if len(lengths) < 2:
        print(f"Слишком мало предложений в файле {clean_file} для расчёта дисперсии.")
        return
    var = statistics.variance(lengths)
    print(f"Дисперсия длины предложений в рассказах {author_name}: {var:.2f}")

# === Применяем ко всем трём авторам ===
print("--- Расчёт дисперсии длины предложений ---")
variance_sentence_length('dostoevsky_clean.txt', 'Достоевского')
variance_sentence_length('garshin_clean.txt', 'Гаршина')
variance_sentence_length('andreev_clean.txt', 'Андреева')
```

###Подсчет частей речи

```python
import os
import re
from natasha import Segmenter, MorphVocab, NewsEmbedding, NewsMorphTagger, Doc

segmenter = Segmenter()
morph_vocab = MorphVocab()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)

def pos_counts(input_file, author_name):
    """
    Подсчитывает процентное соотношение существительных, глаголов и прилагательных.
    """
    if not os.path.exists(input_file):
        print(f"{input_file} не найден. Сначала загрузите файл.")
        return

    with open(input_file, 'r', encoding='utf-8') as f:
        text = f.read()

    doc = Doc(text)
    doc.segment(segmenter)
    doc.tag_morph(morph_tagger)

    total = 0
    counts = {'NOUN': 0, 'VERB': 0, 'ADJ': 0}
    noun_tags = {'NOUN'}
    verb_tags = {'VERB', 'INFN'}
    adj_tags = {'ADJ', 'ADJF', 'ADJS'}

    for token in doc.tokens:
        if token.pos in ('PUNCT', 'SYM', 'X'):
            continue
        total += 1
        if token.pos in noun_tags:
            counts['NOUN'] += 1
        elif token.pos in verb_tags:
            counts['VERB'] += 1
        elif token.pos in adj_tags:
            counts['ADJ'] += 1

    if total == 0:
        print("Нет слов для анализа.")
        return

    result = {k: (v / total * 100) for k, v in counts.items()}
    print(f"--- Части речи в рассказах {author_name} ---")
    print(f"Существительные: {result.get('NOUN', 0):.2f}%")
    print(f"Глаголы: {result.get('VERB', 0):.2f}%")
    print(f"Прилагательные: {result.get('ADJ', 0):.2f}%")

# === Применяем ко всем трём авторам ===
print("Расчёт частей речи")
pos_counts('dostoevsky_clean.txt', 'Достоевского')
pos_counts('garshin_clean.txt', 'Гаршина')
pos_counts('andreev_clean.txt', 'Андреева')
```

###Расчет лексического разнообразия (TTR)

```python
import os

def ttr(lemmas_file, author_name):
    """
    Вычисляем лексическое разнообразие (Type-Token Ratio) по файлу с леммами.
    """
    if not os.path.exists(lemmas_file):
        print(f"{lemmas_file} не найден. Сначала выполните предобработку.")
        return
    with open(lemmas_file, 'r', encoding='utf-8') as f:
        lemmas = f.read().split()
    total = len(lemmas)
    if total == 0:
        print(f"В файле {lemmas_file} нет лемм.")
        return
    unique = len(set(lemmas))
    ttr_value = unique / total
    print(f"Лексическое разнообразие (TTR) у {author_name}: {ttr_value:.4f} (уникальных {unique} из {total})")

# === Применяем ко всем трём авторам ===
print("--- Расчёт лексического разнообразия (TTR) ---")
ttr('dostoevsky_lemmas.txt', 'Достоевского')
ttr('garshin_lemmas.txt', 'Гаршина')
ttr('andreev_lemmas.txt', 'Андреева')
```

###Индекс читаемости

```python
import os
import re

def readability_index(input_file, clean_file, author_name):
    """
    Рассчитываем индекс читаемости Флеша для русского языка.
    """
    if not os.path.exists(input_file):
        print(f"{input_file} не найден. Сначала загрузите файл.")
        return
    if not os.path.exists(clean_file):
        print(f"{clean_file} не найден. Сначала выполните предобработку.")
        return

    with open(input_file, 'r', encoding='utf-8') as f:
        text = f.read()

    words = re.findall(r'[а-яё]+', text.lower())
    if not words:
        print("Слова не найдены.")
        return

    vowels = set('аеёиоуыэюя')
    total_vowels = sum(sum(1 for ch in w if ch in vowels) for w in words)
    total_words = len(words)

    with open(clean_file, 'r', encoding='utf-8') as f:
        sentences = f.read().splitlines()
    total_sentences = len([s for s in sentences if s.strip()])

    if total_sentences == 0 or total_words == 0:
        print("Недостаточно данных для расчёта (ноль слов или предложений).")
        return

    avg_syllables_per_word = total_vowels / total_words
    avg_words_per_sentence = total_words / total_sentences

    readability = 206.835 - (1.3 * avg_words_per_sentence) - (60.1 * avg_syllables_per_word)

    print(f"Индекс читаемости (Флеш) для {author_name}: {readability:.2f}")

# === Применяем ко всем трём авторам ===
print("--- Расчёт индекса читаемости ---")
readability_index('dostoevsky_all.txt', 'dostoevsky_clean.txt', 'Достоевского')
readability_index('garshin_all_clean.txt', 'garshin_clean.txt', 'Гаршина')
readability_index('andreev_all.txt', 'andreev_clean.txt', 'Андреева')
```

```python
import os
import re
import statistics
from natasha import Segmenter, MorphVocab, NewsEmbedding, NewsMorphTagger, Doc
import pandas as pd
from tabulate import tabulate

# Инициализация Natasha (один раз)
segmenter = Segmenter()
morph_vocab = MorphVocab()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)

def compute_all_metrics(input_file, clean_file, lemmas_file, author_name):
    """
    Вычисляем все метрики для одного автора и возвращает словарь.
    """
    metrics = {}

    # 1. Средняя длина предложения (по знакам препинания) – идентично отдельной функции
    if os.path.exists(input_file):
        with open(input_file, 'r', encoding='utf-8') as f:
            text = f.read()
        # Разбиваем текст на предложения по . ! ? (с учётом последующего пробела или конца строки)
        sentences = re.split(r'[.!?]\s+', text)
        # Удаляем пустые или слишком короткие фрагменты (менее 3 символов)
        sentences = [s.strip() for s in sentences if len(s.strip()) > 3]
        # Считаем количество слов в каждом предложении
        word_counts = []
        for sent in sentences:
            words = re.findall(r'[а-яёa-z]+', sent, re.IGNORECASE)
            if words:
                word_counts.append(len(words))
        if word_counts:
            metrics['Средняя длина предложения'] = sum(word_counts) / len(word_counts)
        else:
            metrics['Средняя длина предложения'] = None
    else:
        metrics['Средняя длина предложения'] = None

    # 2. Средняя длина слова
    if os.path.exists(lemmas_file):
        with open(lemmas_file, 'r', encoding='utf-8') as f:
            lemmas = f.read().split()
        metrics['Средняя длина слова'] = sum(len(w) for w in lemmas) / len(lemmas) if lemmas else None
    else:
        metrics['Средняя длина слова'] = None

    # 3. Дисперсия длины предложений (на основе clean_file)
    if os.path.exists(clean_file):
        with open(clean_file, 'r', encoding='utf-8') as f:
            lines = f.read().splitlines()
        lengths = [len(line.split()) for line in lines if line.strip()]
        metrics['Дисперсия длины предложений'] = statistics.variance(lengths) if len(lengths) >= 2 else None
    else:
        metrics['Дисперсия длины предложений'] = None

    # 4. Части речи
    if os.path.exists(input_file):
        with open(input_file, 'r', encoding='utf-8') as f:
            text = f.read()
        doc = Doc(text)
        doc.segment(segmenter)
        doc.tag_morph(morph_tagger)
        total = 0
        counts = {'NOUN': 0, 'VERB': 0, 'ADJ': 0}
        noun_tags = {'NOUN'}
        verb_tags = {'VERB', 'INFN'}
        adj_tags = {'ADJ', 'ADJF', 'ADJS'}
        for token in doc.tokens:
            if token.pos in ('PUNCT', 'SYM', 'X'):
                continue
            total += 1
            if token.pos in noun_tags:
                counts['NOUN'] += 1
            elif token.pos in verb_tags:
                counts['VERB'] += 1
            elif token.pos in adj_tags:
                counts['ADJ'] += 1
        if total > 0:
            metrics['Существительные (%)'] = counts['NOUN'] / total * 100
            metrics['Глаголы (%)'] = counts['VERB'] / total * 100
            metrics['Прилагательные (%)'] = counts['ADJ'] / total * 100
        else:
            metrics['Существительные (%)'] = None
            metrics['Глаголы (%)'] = None
            metrics['Прилагательные (%)'] = None
    else:
        metrics['Существительные (%)'] = None
        metrics['Глаголы (%)'] = None
        metrics['Прилагательные (%)'] = None

    # 5. TTR
    if os.path.exists(lemmas_file):
        with open(lemmas_file, 'r', encoding='utf-8') as f:
            lemmas = f.read().split()
        total = len(lemmas)
        metrics['TTR'] = len(set(lemmas)) / total if total > 0 else None
    else:
        metrics['TTR'] = None

    # 6. Индекс читаемости
    if os.path.exists(input_file) and os.path.exists(clean_file):
        with open(input_file, 'r', encoding='utf-8') as f:
            text = f.read()
        words = re.findall(r'[а-яё]+', text.lower())
        if words:
            vowels = set('аеёиоуыэюя')
            total_vowels = sum(sum(1 for ch in w if ch in vowels) for w in words)
            total_words = len(words)
            with open(clean_file, 'r', encoding='utf-8') as f:
                sentences = f.read().splitlines()
            total_sentences = len([s for s in sentences if s.strip()])
            if total_sentences > 0 and total_words > 0:
                avg_syllables = total_vowels / total_words
                avg_words_sent = total_words / total_sentences
                readability = 206.835 - 1.3 * avg_words_sent - 60.1 * avg_syllables
                metrics['Индекс читаемости (Флеш)'] = readability
            else:
                metrics['Индекс читаемости (Флеш)'] = None
        else:
            metrics['Индекс читаемости (Флеш)'] = None
    else:
        metrics['Индекс читаемости (Флеш)'] = None

    return metrics

# === Список авторов и их файлов ===
authors_info = [
    {'name': 'Достоевский', 'input': 'dostoevsky_all.txt', 'clean': 'dostoevsky_clean.txt', 'lemmas': 'dostoevsky_lemmas.txt'},
    {'name': 'Гаршин',     'input': 'garshin_all_clean.txt', 'clean': 'garshin_clean.txt', 'lemmas': 'garshin_lemmas.txt'},
    {'name': 'Андреев',    'input': 'andreev_all.txt', 'clean': 'andreev_clean.txt', 'lemmas': 'andreev_lemmas.txt'}
]

# === Сбор данных ===
data = {}
for a in authors_info:
    print(f"Обработка {a['name']}...")
    data[a['name']] = compute_all_metrics(a['input'], a['clean'], a['lemmas'], a['name'])

# === Создание DataFrame (авторы — столбцы, метрики — строки) ===
df = pd.DataFrame(data).T  # транспонируем
df = df.T                   # теперь метрики — индекс, авторы — столбцы
df = df.round(2)            # округление до 2 знаков

# === Вывод красивой таблицы с линиями ===
print("\nСводная таблица метрик")
print(tabulate(df, headers='keys', tablefmt='grid', floatfmt='.2f'))

# === Сохранение в CSV ===
df.to_csv('metrics_summary_transposed.csv', encoding='utf-8-sig')
print("\nТаблица сохранена в файл metrics_summary_transposed.csv")
```

###Дельта Берроуза

```python
import numpy as np
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import squareform

# загрузка текста


def load_lemmas(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        return f.read().split()


authors = {
    "Достоевский": "dostoevsky_lemmas.txt",
    "Гаршин": "garshin_lemmas.txt",
    "Андреев": "andreev_lemmas.txt"
}

author_words = {}

for name, path in authors.items():
    words = load_lemmas(path)
    author_words[name] = words
    print(f"{name}: {len(words)} слов")


# частоты слов


def get_word_frequencies(words):
    counter = Counter(words)
    total = len(words)
    return {w: c / total for w, c in counter.items()}


# 150 самых частых слов ВСЕГО корпуса


all_words = []

for words in author_words.values():
    all_words.extend(words)

most_common_words = [
    w for w, _ in Counter(all_words).most_common(150)
]


# частотные векторы


freq_vectors = {}

for author, words in author_words.items():

    freq = get_word_frequencies(words)

    vec = np.array([
        freq.get(word, 0)
        for word in most_common_words
    ])

    freq_vectors[author] = vec


# общие средние и стандартные отклонения

matrix = np.array(list(freq_vectors.values()))

mean = matrix.mean(axis=0)

std = matrix.std(axis=0)

std[std == 0] = 1


# z-оценки

z_vectors = {}

for author in freq_vectors:

    z_vectors[author] = (
        freq_vectors[author] - mean
    ) / std


# Delta Burrows

author_names = list(authors.keys())

n = len(author_names)

delta_matrix = np.zeros((n, n))

for i in range(n):

    for j in range(i + 1, n):

        delta = np.mean(
            np.abs(
                z_vectors[author_names[i]]
                - z_vectors[author_names[j]]
            )
        )

        delta_matrix[i, j] = delta
        delta_matrix[j, i] = delta


df_delta = pd.DataFrame(
    delta_matrix,
    index=author_names,
    columns=author_names
)

print(df_delta.round(4))


# Heatmap


plt.figure(figsize=(6,5))

sns.heatmap(
    df_delta,
    annot=True,
    cmap="RdYlGn_r",
    fmt=".3f",
    square=True
)

plt.title("Burrows' Delta")
plt.show()


# Дендрограмма


condensed = squareform(delta_matrix)

linkage_matrix = linkage(condensed, method="average")

plt.figure(figsize=(7,5))

dendrogram(
    linkage_matrix,
    labels=author_names
)

plt.title("Кластеризация авторов")
plt.show()


# Попарное сравнение


print("\nПопарные значения Delta:\n")

for i in range(n):

    for j in range(i + 1, n):

        print(
            f"{author_names[i]} — {author_names[j]} : {delta_matrix[i,j]:.4f}"
        )
```

```python
#%pip install pymorphy3
from pathlib import Path
from collections import Counter
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from sklearn.metrics.pairwise import cosine_distances

import pymorphy3

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 4)
pd.set_option('display.max_colwidth', 120)


sent_dict = {}

with open("/content/rusentilex_2017.txt", "r", encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(",")

        if len(parts) >= 4:
            word = parts[0].strip().lower()
            sentiment = parts[3].strip().lower()

            sent_dict[word] = sentiment

files = {
    "Достоевский": "dostoevsky_lemmas.txt",
    "Гаршин": "garshin_lemmas.txt",
    "Андреев": "andreev_lemmas.txt"
}

def score_lemmas(lemmas):
    score = 0
    total = 0

    for w in lemmas:
        w = w.lower()

        if w in sent_dict:
            total += 1

            if sent_dict[w] == "positive":
                score += 1
            elif sent_dict[w] == "negative":
                score -= 1

    return score / total if total > 0 else 0


def sentiment_curve_for_file(path, points=100):
    text = Path(path).read_text(encoding='utf-8')

    lemmas = text.split()

    chunks = np.array_split(lemmas, points)

    scores = np.array([score_lemmas(chunk) for chunk in chunks])

    return scores


curves = {}

for name, file in files.items():
    print("Обрабатываю:", name)
    curves[name] = sentiment_curve_for_file(file)

curve_df = pd.DataFrame(curves)

plt.figure(figsize=(11, 5))

for name in curve_df.columns:
    plt.plot(curve_df.index, curve_df[name], label=name)

plt.title("Эмоциональные кривые")
plt.xlabel("Позиция текста (0–100)")
plt.ylabel("Тональность")
plt.legend()
plt.show()
```

```python
import re
import numpy as np
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.cluster.hierarchy import dendrogram, linkage

# ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ


def get_word_frequencies(words):
    total = len(words)
    if total == 0:
        return {}
    counter = Counter(words)
    return {w: count / total for w, count in counter.items()}

def split_text_into_chunks(words, chunk_size=500):
    return [words[i:i+chunk_size] for i in range(0, len(words), chunk_size) if len(words[i:i+chunk_size]) >= chunk_size]

def load_lemmas(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read().split()


# ПОСТРОЕНИЕ ОБЩЕГО КОРПУСА И MFW


def build_mfw(chunks, n_most_common=150):
    counter = Counter()
    for chunk in chunks:
        counter.update(chunk)
    return [w for w, _ in counter.most_common(n_most_common)]

def build_corpus_statistics(chunks, mfw):
    matrix = []
    for chunk in chunks:
        freq = get_word_frequencies(chunk)
        matrix.append([freq.get(w, 0) for w in mfw])
    matrix = np.array(matrix)
    mean = matrix.mean(axis=0)
    std = matrix.std(axis=0)
    std[std == 0] = 1e-9
    return mean, std


# КЛАССИЧЕСКАЯ ДЕЛЬТА БЁРРОУЗА

def burrows_delta(words1, words2, mfw, corpus_mean, corpus_std):
    freq1 = get_word_frequencies(words1)
    freq2 = get_word_frequencies(words2)
    vec1 = np.array([freq1.get(w, 0) for w in mfw])
    vec2 = np.array([freq2.get(w, 0) for w in mfw])
    z1 = (vec1 - corpus_mean) / corpus_std
    z2 = (vec2 - corpus_mean) / corpus_std
    return np.mean(np.abs(z1 - z2))


# РАСЧЁТ ВНУТРИАВТОРСКОЙ ВАРИАТИВНОСТИ

def calculate_author_variability(words, chunk_size, mfw, corpus_mean, corpus_std):
    chunks = split_text_into_chunks(words, chunk_size)
    if len(chunks) < 2:
        return None, None, None, chunks
    n = len(chunks)
    deltas = []
    for i in range(n):
        for j in range(i+1, n):
            d = burrows_delta(chunks[i], chunks[j], mfw, corpus_mean, corpus_std)
            deltas.append(d)
    mean_delta = np.mean(deltas)
    std_delta = np.std(deltas)
    return mean_delta, std_delta, chunks

# ОСНОВНАЯ ПРОГРАММА

print("=" * 70)
print("КЛАССИЧЕСКАЯ ДЕЛЬТА БЁРРОУЗА (ВНУТРИАВТОРСКАЯ ВАРИАТИВНОСТЬ)")
print("=" * 70)

authors = {
    'Достоевский': 'dostoevsky_lemmas.txt',
    'Гаршин':      'garshin_lemmas.txt',
    'Андреев':     'andreev_lemmas.txt'
}

author_words = {}
for name, file_path in authors.items():
    try:
        words = load_lemmas(file_path)
        author_words[name] = words
        print(f"Загружено {name}: {len(words)} слов")
    except FileNotFoundError:
        print(f"ОШИБКА: файл {file_path} не найден")
        author_words[name] = []

if any(len(w) == 0 for w in author_words.values()):
    print("Некоторые файлы не загружены. Проверьте наличие файлов.")
    exit()

CHUNK_SIZE = 500
all_chunks = []
for words in author_words.values():
    all_chunks.extend(split_text_into_chunks(words, CHUNK_SIZE))

if len(all_chunks) < 2:
    raise ValueError("Недостаточно чанков для построения корпуса.")

print(f"Всего чанков в корпусе: {len(all_chunks)}")

MFW = build_mfw(all_chunks, n_most_common=150)
CORPUS_MEAN, CORPUS_STD = build_corpus_statistics(all_chunks, MFW)
print(f"Размер MFW: {len(MFW)} слов")

results = {}
for name, words in author_words.items():
    mean_d, std_d, chunks = calculate_author_variability(
        words, CHUNK_SIZE, MFW, CORPUS_MEAN, CORPUS_STD
    )
    if mean_d is None:
        print(f"У {name} слишком мало чанков для расчёта (нужно >= 2)")
        continue
    results[name] = {
        'mean_delta': mean_d,
        'std_delta': std_d,
        'num_chunks': len(chunks)
    }

# --- Таблица результатов ---
print("\n" + "=" * 70)
print("СРАВНИТЕЛЬНАЯ ТАБЛИЦА")
print("=" * 70)

df = pd.DataFrame.from_dict(results, orient='index')
df['similarity'] = 1 / (1 + df['mean_delta'])
df = df[['mean_delta', 'std_delta', 'num_chunks', 'similarity']]
df.columns = ['Средняя дельта', 'Стд. отклонение', 'Кол-во чанков', 'Нормированный индекс сходства']
print(df.round(4))

# --- Визуализация индекса сходства ---
plt.figure(figsize=(10, 5))
bars = plt.bar(df.index, df['Нормированный индекс сходства'], color=['#3498db', '#2ecc71', '#e74c3c'])
plt.title('Нормированный индекс сходства (1/(1+δ))', fontsize=14)
plt.ylabel('Индекс сходства (чем выше, тем гомогеннее стиль)', fontsize=12)
plt.ylim(0, 1)
plt.grid(axis='y', alpha=0.3)
for bar, val in zip(bars, df['Нормированный индекс сходства']):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01,
             f'{val:.3f}', ha='center', fontsize=10)
plt.tight_layout()
plt.show()

# --- Ранжирование и сравнение (исправлено) ---
ranking = sorted(results.items(), key=lambda x: x[1]['mean_delta'])

print("\n" + "=" * 70)
print("РАНЖИРОВАНИЕ АВТОРОВ ПО ВНУТРЕННЕЙ ВАРИАТИВНОСТИ (от низкой к высокой)")
print("=" * 70)
for i, (name, data) in enumerate(ranking, 1):
    print(f"{i}. {name:15s} : средняя дельта = {data['mean_delta']:.4f}  ({data['num_chunks']} чанков)")

print("\n" + "=" * 70)
print("СРАВНЕНИЕ ПОПАРНО")
print("=" * 70)
for i in range(len(ranking)-1):
    name1, data1 = ranking[i]
    name2, data2 = ranking[i+1]
    diff = (data2['mean_delta'] - data1['mean_delta']) / data1['mean_delta'] * 100
    print(f"  {name1} ({data1['mean_delta']:.3f}) → {name2} ({data2['mean_delta']:.3f}) : "
          f"дельта увеличивается на {diff:.1f}%")


# ДОПОЛНИТЕЛЬНАЯ ВИЗУАЛИЗАЦИЯ: МЕЖАВТОРСКАЯ МАТРИЦА ДЕЛЬТ

print("\n" + "=" * 70)
print("ПОСТРОЕНИЕ МАТРИЦЫ МЕЖАВТОРСКИХ ДЕЛЬТ (карта гомогенности)")
print("=" * 70)

# Собираем по 10 чанков от каждого автора для наглядности
sample_chunks = []
sample_labels = []
for name, words in author_words.items():
    chunks = split_text_into_chunks(words, CHUNK_SIZE)[:10]
    sample_chunks.extend(chunks)
    sample_labels.extend([name] * len(chunks))

n_chunks = len(sample_chunks)
delta_between_chunks = np.zeros((n_chunks, n_chunks))
for i in range(n_chunks):
    for j in range(i+1, n_chunks):
        d = burrows_delta(sample_chunks[i], sample_chunks[j], MFW, CORPUS_MEAN, CORPUS_STD)
        delta_between_chunks[i, j] = d
        delta_between_chunks[j, i] = d

# Тепловая карта
plt.figure(figsize=(12, 10))
sns.heatmap(delta_between_chunks,
            xticklabels=sample_labels,
            yticklabels=sample_labels,
            cmap='RdYlGn_r',
            square=True,
            cbar_kws={'label': 'Дельта Бёрроуза'})
plt.title('Тепловая карта стилистических расстояний между чанками (авторы отмечены цветом)', fontsize=14)
plt.xlabel('Чанк (автор)', fontsize=12)
plt.ylabel('Чанк (автор)', fontsize=12)
plt.tight_layout()
plt.show()

# Дендрограмма
print("\n" + "=" * 70)
print("ДЕНДРОГРАММА СТИЛИСТИЧЕСКОЙ БЛИЗОСТИ ЧАНКОВ")
print("=" * 70)

linkage_matrix = linkage(delta_between_chunks, method='average')
plt.figure(figsize=(14, 6))
dendrogram(linkage_matrix,
           labels=sample_labels,
           orientation='top',
           leaf_rotation=90,
           leaf_font_size=10,
           color_threshold=np.mean(linkage_matrix[:, 2]) * 0.5)
plt.title('Дендрограмма стилистической близости чанков', fontsize=14)
plt.xlabel('Чанки (с указанием автора)', fontsize=12)
plt.ylabel('Расстояние (дельта Бёрроуза)', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
from scipy.cluster.hierarchy import dendrogram, linkage, leaves_list
from scipy.spatial.distance import squareform   # <--- ДОБАВЛЕНО


# ВСПОМОГАТЕЛЬНЫЕ ФУНКЦИИ

def get_word_frequencies(words):
    total = len(words)
    if total == 0:
        return {}
    counter = Counter(words)
    return {w: count / total for w, count in counter.items()}

def split_text_into_chunks(words, chunk_size=500):
    return [words[i:i+chunk_size] for i in range(0, len(words), chunk_size) if len(words[i:i+chunk_size]) >= chunk_size]

def load_lemmas(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        return f.read().split()

def build_mfw(chunks, n_most_common=150):
    counter = Counter()
    for chunk in chunks:
        counter.update(chunk)
    return [w for w, _ in counter.most_common(n_most_common)]

def build_corpus_statistics(chunks, mfw):
    matrix = []
    for chunk in chunks:
        freq = get_word_frequencies(chunk)
        matrix.append([freq.get(w, 0) for w in mfw])
    matrix = np.array(matrix)
    mean = matrix.mean(axis=0)
    std = matrix.std(axis=0)
    std[std == 0] = 1e-9
    return mean, std

def burrows_delta(words1, words2, mfw, corpus_mean, corpus_std):
    freq1 = get_word_frequencies(words1)
    freq2 = get_word_frequencies(words2)
    vec1 = np.array([freq1.get(w, 0) for w in mfw])
    vec2 = np.array([freq2.get(w, 0) for w in mfw])
    z1 = (vec1 - corpus_mean) / corpus_std
    z2 = (vec2 - corpus_mean) / corpus_std
    return np.mean(np.abs(z1 - z2))

# ПОСТРОЕНИЕ КАРТЫ ГОМОГЕННОСТИ ДЛЯ ОДНОГО АВТОРА

def plot_author_heatmap(author_name, words, mfw, corpus_mean, corpus_std,
                        chunk_size=500, max_chunks=30):
    """
    Строит тепловую карту попарных дельт между чанками одного автора.
    """
    chunks = split_text_into_chunks(words, chunk_size)
    if len(chunks) > max_chunks:
        chunks = chunks[:max_chunks]
        print(f"{author_name}: взято {max_chunks} чанков (из {len(chunks)} существующих)")
    else:
        print(f"{author_name}: всего {len(chunks)} чанков")

    if len(chunks) < 2:
        print(f"{author_name}: недостаточно чанков для построения карты")
        return

    n = len(chunks)
    delta_matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(i+1, n):
            d = burrows_delta(chunks[i], chunks[j], mfw, corpus_mean, corpus_std)
            delta_matrix[i, j] = d
            delta_matrix[j, i] = d

    # --- ИСПРАВЛЕНО: используем squareform для правильной кластеризации ---
    linkage_matrix = linkage(squareform(delta_matrix), method='average')
    order = leaves_list(linkage_matrix)
    delta_ordered = delta_matrix[order, :][:, order]

    # Тепловая карта
    plt.figure(figsize=(10, 8))
    sns.heatmap(delta_ordered,
                xticklabels=False, yticklabels=False,
                cmap='RdYlGn_r',
                square=True,
                cbar_kws={'label': 'Дельта Бёрроуза'})
    plt.title(f'Карта гомогенности стиля: {author_name}\n(Чанки по {chunk_size} слов, показано {n} чанков)',
              fontsize=14)
    plt.xlabel('Чанк')
    plt.ylabel('Чанк')
    plt.tight_layout()
    plt.show()

    # Дендрограмма (тоже с squareform)
    plt.figure(figsize=(12, 4))
    dendrogram(linkage_matrix,
               labels=[f'#{i}' for i in range(n)],
               orientation='top',
               leaf_rotation=90,
               leaf_font_size=8)
    plt.title(f'Дендрограмма чанков: {author_name}')
    plt.xlabel('Чанк')
    plt.ylabel('Стилистическое расстояние (дельта)')
    plt.tight_layout()
    plt.show()


# ОСНОВНАЯ ПРОГРАММА

print("=" * 70)
print("ПОСТРОЕНИЕ ТРЁХ КАРТ ГОМОГЕННОСТИ (ДОСТОЕВСКИЙ, ГАРШИН, АНДРЕЕВ)")
print("=" * 70)

# --- 1. Загрузка файлов лемм ---
authors = {
    'Достоевский': 'dostoevsky_lemmas.txt',
    'Гаршин':      'garshin_lemmas.txt',
    'Андреев':     'andreev_lemmas.txt'
}

author_words = {}
for name, path in authors.items():
    try:
        author_words[name] = load_lemmas(path)
        print(f"Загружено {name}: {len(author_words[name])} слов")
    except FileNotFoundError:
        print(f"ОШИБКА: файл {path} не найден")
        author_words[name] = []

if any(len(w) == 0 for w in author_words.values()):
    print("Некоторые файлы не загружены. Проверьте наличие файлов.")
    exit()

# --- 2. Построение общего корпуса чанков и глобальных статистик ---
CHUNK_SIZE = 500
all_chunks = []
for words in author_words.values():
    all_chunks.extend(split_text_into_chunks(words, CHUNK_SIZE))

print(f"Всего чанков в корпусе: {len(all_chunks)}")

MFW = build_mfw(all_chunks, n_most_common=150)
CORPUS_MEAN, CORPUS_STD = build_corpus_statistics(all_chunks, MFW)
print(f"Размер MFW: {len(MFW)} слов")

# --- 3. Построение карт для каждого автора ---
for name, words in author_words.items():
    plot_author_heatmap(name, words, MFW, CORPUS_MEAN, CORPUS_STD,
                        chunk_size=CHUNK_SIZE, max_chunks=30)
```

Обнаруженные различия отражены практически в большинстве выбранных характеристик. А именно в следующих: средняя длина предложения, дисперсия длины предложений, индекс читаемости, Дельта Берроуза, TF-IDF.

Более подробно рассмотрим, в чём именно заключаются отличия стиля Л. Н. Андреева по каждой из перечисленных характеристик. Самый большой показатель средней длины предложения является значимой отличительной чертой стиля автора. К этому параметру стоит добавить и наибольшую дисперсию длин предложений у Л. Н. Андреева среди других писателей. Это значит, что автор использует значительно более длинные синтаксические конструкции*, а предложения в его тексте сильно отличаются друг от друга по количеству слов.

*Здесь укажем, что в малой прозе Ф. М. Достоевского зачастую встречаются диалоги. В том числе с репликами, состоящими из одного-двух слов. Не исключено, что такая специфика текста может быть связана с построчной оплатой произведения писателю. Тем не менее, этот фактор, вероятноменьшая его реальное значение.  

Другой важной отличительной чертой стиля Л. Н. Андреева является индекс читаемости. У писателя наблюдается наиболее низкий индекс читаемости, то есть его тексты оказываются для читателя наиболее трудными, что может быть связано и вариативностью длин предложений и синтаксиса.

Рассматривая стиль Л. Н. Андреева в контексте предыдущей литературной традиции, ценными выглядят результаты Дельты Берроуза. Согласно её показателям, Л. Н. Андреев и Ф. М. Достоевский наиболее удалены друг от друга, а ближе всего друг другу оказываются Л. Н. Андреев и В. М. Гаршин.

Некоторые показатели в стилях авторов сходятся. А именно у трёх писателей практически одинакова средняя длина слова и количественное соотношение частей речи (имён прилагательных, имён существительных и глаголов) в их текстах. Данные наблюдения подтверждают, что стилевые различия между авторами обусловлены не простым использованием более тяжеловесных слов или бОльшим количеством конкретной части речи, а более сложной организацией текста.

Также результаты меры TF-IDF через выдачу самых популярных слов корпусов показывают как сходства стилей авторов, так и их различия. Например, и Л. Н. Андреев, и В. М. Гаршин очень физиологичны и часто используют слова "рука", "глаз", "лицо", "голова". Но несмотря на схожие темы произведений писателей, только у Л. Н. Андреева, чьи тексты наполнены мотивами ужаса и безумия, встречаются слова "страшный", "ночь" и "черный". Так, например, отличительной чертой корпуса Ф. М. Достоевского можно назвать слово "Бог", которое оказывается в топе по количеству употреблений.  

Заметим, что некоторые результаты исследования стоит использовать с осторожностью. Например, метрика TTR обозначает лексическое разгнообразие Л. Н. Андреева как самое низкое. Но следует учитывать, что данный показатель зависит от объёма анализируемого текста. Вероятно, наблюдаемое снижение словарного запаса связано не только со стилем автора, но и с существенно большим объёмом его корпуса, в отличие от корпуса Ф. М. Достоевского. К подобным ограничения приходится прибнгнуть при рассмотрении эмоциональной тональности текстов, что связано с их разнородностью в корпусах текстов малой прозы.

###Встраивание в научный контекст

Говоря о научном контексте, обратим внимание на то, как полученные выводы соотносятся с филологической традицией. В литературоведении тексты и Л. Н. Андреева, и В. М. Гаршина, и Ф. М. Достоевского связывают с мотивами безумия и используемым методом психологизма. Не единожды сравнивали Л. Н. Андреева и В. М. Гаршина, как авторов, разрабатывающих тему сумасшествия в модернизме. В том числе, рассматривая наследование авторов литературной традиции Ф. М. Достоевского.

Полученные нами в ходе исследования результаты согласуются с классическими представлениями о соотношениях литературы Л. Н. Андреева, Ф. М. Достоевксого и В. М. Гаршина. Но также добавляют в эту дискуссию сопоставление сразу трёх авторов, связанных едиными темами, и последовательно разрабатывающих литературную традицию.  

Также использование цифровых корпусных методов позволило количественно подтвердить конкретные особенности авторского стиля Л. Н. Андреева, которые в литературоведении описываются преимущественно качественными методами анализа.
